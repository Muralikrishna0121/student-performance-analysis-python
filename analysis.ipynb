```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Environment Ready!")
```

```python
df = pd.read_csv("data/StudentsPerformance.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
avg_math = df["math score"].mean()
avg_read = df["reading score"].mean()
avg_write = df["writing score"].mean()

print("Math Average:", avg_math)
print("Reading Average:", avg_read)
print("Writing Average:", avg_write)
```

```python
df["total_score"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
)

df.head()
```

```python
top_students = df.sort_values(
    by="total_score",
    ascending=False
).head(5)

top_students
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["math score"], bins=20)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Count")

plt.savefig("visuals/chart1.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df["math score"], bins=20)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Count")

plt.savefig("visuals/chart1.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
avg_gender = df.groupby("gender")[["math score","reading score","writing score"]].mean()

avg_gender.plot(kind="bar", figsize=(10,5))

plt.title("Average Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")

plt.savefig("visuals/chart2.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
parent_avg = df.groupby("parental level of education")["math score"].mean()

parent_avg.plot(kind="bar", figsize=(12,5))

plt.title("Parental Education vs Math Score")
plt.xlabel("Parental Education")
plt.ylabel("Average Math Score")

plt.savefig("visuals/chart3.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
race_avg = df.groupby("race/ethnicity")["math score"].mean()

race_avg.plot(kind="bar", figsize=(10,5))

plt.title("Race/Ethnicity vs Average Math Score")
plt.xlabel("Race/Ethnicity")
plt.ylabel("Average Math Score")

plt.savefig("visuals/chart4.png", dpi=300, bbox_inches="tight")

plt.show()
```

1. Reading scores are generally higher than math scores.

5. Group E students achieve the highest average scores.

6. Top-performing students score consistently across all subjects.

8. Writing and reading scores are strongly related.

9. Academic performance varies across demographic groups.
